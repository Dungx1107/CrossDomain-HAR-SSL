```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Cấu hình hiển thị đồ thị
plt.style.use("seaborn-v0_8-whitegrid" if "seaborn-v0_8-whitegrid" in plt.style.available else "default")
plt.rcParams["figure.figsize"] = (14, 6)
```

```python
DATA_DIR_MOTION_SENSE = Path("../../data/raw/motion_sense")
print(f"Kiểm tra thư mục dữ liệu: {DATA_DIR_MOTION_SENSE.exists()}")
```

```python
folder_motion_sense = Path(DATA_DIR_MOTION_SENSE)
print(folder_motion_sense)
for con in folder_motion_sense.iterdir():
    if con.is_dir():
        files = list(con.glob("*.csv"))
        print(con, ": ", len(files))
```

```python
folder_walk = folder_motion_sense.glob("wlk_*")
print(folder_walk)
for i in folder_walk:
    file_csv = list(i.glob("*.csv"))
    print(i, " has ", len(file_csv), " files csv")
    for j in file_csv:
        print(j)
```

```python
wlk8_sub1 = "../../data/raw/motion_sense/wlk_8/sub_1.csv"
wlk8_sub2 = "../../data/raw/motion_sense/wlk_8/sub_2.csv"

df_wlk8_sub1 = pd.read_csv(wlk8_sub1)
df_wlk8_sub2 = pd.read_csv(wlk8_sub2)

print(df_wlk8_sub1.info())
print(df_wlk8_sub2.info())
```

```python

if "Unnamed: 0" in df_wlk8_sub1.columns:
    df_wlk8_sub1 = df_wlk8_sub1.drop("Unnamed: 0", axis=1)
    df_wlk8_sub2 = df_wlk8_sub2.drop("Unnamed: 0", axis=1)

plt.figure(figsize=(30, 10))

for cols in df_wlk8_sub1.columns:
    plt.plot(df_wlk8_sub1[cols][:250], label=cols)

plt.legend()
plt.show()
```

```python
plt.figure(figsize=(30, 10))
for cols in df_wlk8_sub1.columns:
    plt.plot(df_wlk8_sub1[cols][:250], label=cols)


plt.legend()
plt.show()
```

```python
# Tạo danh sách các nhóm kênh
attitude_cols = ["attitude.roll", "attitude.pitch", "attitude.yaw"]
gravity_cols = ["gravity.x", "gravity.y", "gravity.z"]
rotation_cols = ["rotationRate.x", "rotationRate.y", "rotationRate.z"]
user_acc_cols = ["userAcceleration.x", "userAcceleration.y", "userAcceleration.z"]
```

```python
plt.rcParams['axes.titlesize'] = 34  # Cỡ chữ title
plt.rcParams['axes.labelsize'] = 18  # Cỡ chữ label trục
plt.rcParams['legend.fontsize'] = 26  # Cỡ chữ legend

df_wlk8_sub1[attitude_cols][:150].plot(
    figsize=(30, 10),
    title="attitude walk sub1",
    color=['red', 'green', 'blue']
)
plt.legend()
plt.show()

df_wlk8_sub2[attitude_cols][:150].plot(
    figsize=(30, 10),
    title="attitude walk sub2",
    color=['red', 'green', 'blue']
)
plt.legend()
plt.show()

```

```python
df_wlk8_sub1[gravity_cols][:150].plot(
    figsize=(30, 10),
    title="gravity walk sub1",
    color=['red', 'green', 'blue']
)
plt.legend()
plt.show()

df_wlk8_sub2[gravity_cols][:150].plot(
    figsize=(30, 10),
    title="gravity walk sub2",
    color=['red', 'green', 'blue']
)
plt.legend()
plt.show()

```

```python
df_wlk8_sub1[rotation_cols][:150].plot(
    figsize=(30, 10),
    title="rotationRate walk sub1",
    color=['red', 'green', 'blue']
)
plt.legend()
plt.show()

df_wlk8_sub2[rotation_cols][:150].plot(
    figsize=(30, 10),
    title="rotationRate walk sub2",
    color=['red', 'green', 'blue']
)
plt.legend()
plt.show()

```

```python
df_wlk8_sub1[user_acc_cols][:150].plot(
    figsize=(30, 10),
    title="userAcceleration walk sub1",
    color=['red', 'green', 'blue']
)
plt.legend()
plt.show()

df_wlk8_sub2[user_acc_cols][:150].plot(
    figsize=(30, 10),
    title="userAcceleration walk sub2",
    color=['red', 'green', 'blue']
)
plt.legend()
plt.show()
```
