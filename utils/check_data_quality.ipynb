```python
import torch
from pathlib import Path
from config.hhar_config import HHARConfig

WATCH_DIR = Path(HHARConfig.PROCESSED_DIR_WATCH)
SPLITS = ["train.pt", "val.pt", "test.pt", "dataset_all.pt"]

# Danh sách 5 lớp tiêu chuẩn của đề tài
CLASS_NAMES = ['Walking', 'Upstairs', 'Downstairs', 'Sitting', 'Standing']

print("=" * 80)
print(f"{'CHI TIẾT PHÂN PHỐI NHÃN TẬP HHAR_WATCH':^80}")
print("=" * 80)

for split in SPLITS:
    file_path = WATCH_DIR / split
    data = torch.load(file_path, map_location="cpu", weights_only=True)

    labels = data["labels"] if isinstance(data, dict) else data
    if not isinstance(labels, torch.Tensor):
        labels = torch.tensor(labels)
    labels = labels.long()

    unique_classes, counts = torch.unique(labels, return_counts=True)
    count_dict = {int(c): int(cnt) for c, cnt in zip(unique_classes, counts)}

    print(f"\n📂 Split: {split:<14} | Tổng số mẫu: {len(labels)}")
    print(f"{'Label ID':<10} | {'Tên hoạt động':<18} | {'Số lượng mẫu':<15} | {'Tỷ lệ %'}")
    print("-" * 65)

    for idx, name in enumerate(CLASS_NAMES):
        cnt = count_dict.get(idx, 0)
        pct = (cnt / len(labels)) * 100 if len(labels) > 0 else 0.0
        missing_flag = " ⚠️ (MẤT LỚP)" if cnt == 0 else ""
        print(f"{idx:<10} | {name:<18} | {cnt:<15} | {pct:6.2f}%{missing_flag}")

print("\n" * 1 + "=" * 80)
```
